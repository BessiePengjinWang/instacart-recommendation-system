```python
# Cell: Quick Data Quality Check
print("="*60)
print("DATA QUALITY CHECK")
print("="*60)

# Define feature_cols here so it works standalone
feature_cols = [col for col in train_df.columns 
                if col not in ['user_id', 'product_id', 'reordered']]

print(f"\n✅ Train dataset: {train_df.shape}")
print(f"✅ Val dataset: {val_df.shape}")
print(f"✅ Total features: {len(feature_cols)}")
print(f"   Feature list: {feature_cols}")

print("\n📊 Label Distribution:")
print(f"  Train positive rate: {train_df['reordered'].mean():.2%}")
print(f"  Val positive rate: {val_df['reordered'].mean():.2%}")

print("\n🔍 Missing Values:")
missing_train = train_df.isnull().sum().sum()
missing_val = val_df.isnull().sum().sum()
print(f"  Train: {missing_train} missing values")
print(f"  Val: {missing_val} missing values")

if missing_train > 0:
    print("\n  Columns with missing values:")
    missing_cols = train_df.isnull().sum()[train_df.isnull().sum() > 0]
    print(missing_cols)

print("\n✅ Data ready for modeling!" if missing_train == 0 else "\n⚠️  Need to handle missing values")
```
