```python
import pandas as pd
import numpy as np

# Load data
orders = pd.read_csv('../data/raw/orders.csv')
order_products_prior = pd.read_csv('../data/raw/order_products__prior.csv')
order_products_train = pd.read_csv('../data/raw/order_products__train.csv')

# Understand dataset split semantics
print("Eval set distribution:")
print(orders['eval_set'].value_counts())
```

```python
# Step 1: Get all users in train set
train_orders = orders[orders['eval_set'] == 'train']
train_users = train_orders['user_id'].unique()

print(f"Total users in train set: {len(train_users)}")

# Step 2: Split train users into train/val (80/20)
from sklearn.model_selection import train_test_split

train_users_split, val_users_split = train_test_split(
    train_users, test_size=0.2, random_state=42
)

print(f"Train users: {len(train_users_split)}")
print(f"Val users: {len(val_users_split)}")
```
